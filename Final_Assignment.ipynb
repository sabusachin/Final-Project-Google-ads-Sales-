```python
for col in df2.select_dtypes(include="number").columns:
    Q1 = df2[col].quantile(0.25)
    Q3 = df2[col].quantile(0.75)
    IQR = Q3 - Q1

    outliers = df2[
        (df2[col] < Q1 - 1.5 * IQR) |
        (df2[col] > Q3 + 1.5 * IQR)
    ]

    print(col, ":", len(outliers), "outliers")
```

I used the IQR method to find outliers in the dataset. No outliers were found in most columns, but 79 were found in Conversion_Percentage. I kept these values because they may be genuine results.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.boxplot(df2["Conversion_Percentage"].dropna())

plt.title("Conversion Percentage Box Plot")
plt.ylabel("Conversion Percentage")

plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.scatter(df2["Clicks"], df2["Conversions"])

plt.title("Clicks vs Conversions")
plt.xlabel("Clicks")
plt.ylabel("Conversions")

plt.tight_layout()
plt.show()
```

The scatter plot shows the relationship between clicks and conversions. The points are spread across the chart, and there is no clear pattern showing that conversions increase as clicks increase.

```python
import matplotlib.pyplot as plt

device_counts = df2["Device"].value_counts()

plt.figure(figsize=(7, 7))
plt.pie(device_counts,labels=device_counts.index,autopct="%1.1f%%",startangle=90)

plt.title("Ads by Device (%)")
plt.show()
```

The pie chart shows the percentage of ads by device. Desktop has the highest share at 34.3%, followed by Mobile at 33.8% and Tablet at 31.9%. The percentages are very similar.

```python
import seaborn as sns
import matplotlib.pyplot as plt

numeric_data = df2[["Clicks", "Cost", "Leads", "Conversions", "Sale_Amount"]]
correlation = numeric_data.corr()
plt.figure(figsize=(8, 6))
sns.heatmap(correlation, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```

The heatmap shows that changes in clicks, cost, leads, and conversions are not strongly related to each other in this dataset.
